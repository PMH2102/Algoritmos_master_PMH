##Devolución de cambio por técnica voraz

```
def cambio_monedas(N, SM):
  SOLUCION = [0]*len(SM)   #SOLUCION = [0,0,0,0,..]
  ValorAcumulado = 0

  for i,valor in enumerate(SM):
    monedas =  (N-ValorAcumulado)//valor
    SOLUCION[i] = monedas
    ValorAcumulado = ValorAcumulado + monedas*valor

    if ValorAcumulado == N:
      return SOLUCION


cambio_monedas(15,[25,10,5,1])
```

## N-Reinas por técnica de vueta atrás


```
def escribe(S):
  n = len(S)
  for x in range(n):
    print("")
    for i in range(n):
      if S[i] == x+1:
        print(" X " , end="")
      else:
        print(" - ", end="")


def es_prometedora(SOLUCION,etapa):
  #print(SOLUCION)
  #Si la solución tiene dos valores iguales no es valida => Dos reinas en la misma fila
  for i in range(etapa+1):
    #print("El valor " + str(SOLUCION[i]) + " está " +  str(SOLUCION.count(SOLUCION[i])) + " veces")
    if SOLUCION.count(SOLUCION[i]) > 1:
      return False

    #Verifica las diagonales
    for j in range(i+1, etapa +1 ):
      #print("Comprobando diagonal de " + str(i) + " y " + str(j))
      if abs(i-j) == abs(SOLUCION[i]-SOLUCION[j]) : return False
  return True



def reinas(N, solucion=[], etapa=0):
  if len(solucion) == 0:
      solucion=[0 for i in range(N)]

  for i in range(1, N+1):
    solucion[etapa] = i

    if es_prometedora(solucion, etapa):
      if etapa == N-1:
        print(solucion)
        #escribe(solucion)
        print()
      else:
        reinas(N, solucion, etapa+1)
    else:
      None

    solucion[etapa] = 0

reinas(8)
```

## Viaje por el rio. Programación dinámica

```
TARIFAS = [
[0,5,4,3,999,999,999],
[999,0,999,2,3,999,11],
[999,999, 0,1,999,4,10],
[999,999,999, 0,5,6,9],
[999,999, 999,999,0,999,4],
[999,999, 999,999,999,0,3],
[999,999,999,999,999,999,0]
]



################################################################
def Precios(TARIFAS):
################################################################
  #Total de Nodos
  N = len(TARIFAS[0])

  #Inicialización de la tabla de precios
  PRECIOS = [ [9999]*N for i in [9999]*N]
  RUTA = [ [""]*N for i in [""]*N]

  for i in range(0,N-1):
    RUTA[i][i] = i             #Para ir de i a i se "pasa por i"
    PRECIOS[i][i] = 0          #Para ir de i a i se se paga 0
    for j in range(i+1, N):
      MIN = TARIFAS[i][j]
      RUTA[i][j] = i

      for k in range(i, j):
        if PRECIOS[i][k] + TARIFAS[k][j] < MIN:
            MIN = min(MIN, PRECIOS[i][k] + TARIFAS[k][j] )
            RUTA[i][j] = k          #Anota que para ir de i a j hay que pasar por k
        PRECIOS[i][j] = MIN

  return PRECIOS,RUTA
################################################################


PRECIOS,RUTA = Precios(TARIFAS)
#print(PRECIOS[0][6])

print("PRECIOS")
for i in range(len(TARIFAS)):
  print(PRECIOS[i])

print("\nRUTA")
for i in range(len(TARIFAS)):
  print(RUTA[i])

#Determinar la ruta con Recursividad
def calcular_ruta(RUTA, desde, hasta):
  if desde == hasta:
    #print("Ir a :" + str(desde))
    return ""
  else:
    return str(calcular_ruta( RUTA, desde, RUTA[desde][hasta])) +  \
                ',' + \
                str(RUTA[desde][hasta] \
              )

print("\nLa ruta es:")
calcular_ruta(RUTA, 0,6)
```

La complejidad del algoritmo anterior es de n^2 ya que vemos que tiene dos bucles.
Se podría mejorar ordenando primero los números ya que el par más cercano estará siempre consecutivo,entonces en ver de comparar todos solo comparas con los vecinos ya que cuanto más legos de la lista mas distancia habrá y la complejidad cambia a O(n log n).

```

LISTA_1D.sort()

def closest_pair_1d(lista):

    n = len(lista)

    # Caso base: un solo elemento
    if n == 1:
        return None, None, float("inf")

    # Caso base: dos elementos
    if n == 2:
        distancia = abs(lista[1] - lista[0])

        # Ignorar pares con distancia 0
        if distancia == 0:
            return None, None, float("inf")

        return lista[0], lista[1], distancia

    # Dividir en dos mitades
    mitad = n // 2

    izquierda = lista[:mitad]
    derecha = lista[mitad:]

    # Resolver recursivamente cada mitad
    p1_izq, p2_izq, dist_izq = closest_pair_1d(izquierda)
    p1_der, p2_der, dist_der = closest_pair_1d(derecha)

    # Comparar el último de la izquierda con el primero de la derecha
    p1_cruce = izquierda[-1]
    p2_cruce = derecha[0]

    dist_cruce = abs(p2_cruce - p1_cruce)

    # Ignorar distancia 0
    if dist_cruce == 0:
        dist_cruce = float("inf")

    # Elegir la mejor de las tres opciones
    if dist_izq <= dist_der and dist_izq <= dist_cruce:
        return p1_izq, p2_izq, dist_izq

    elif dist_der <= dist_izq and dist_der <= dist_cruce:
        return p1_der, p2_der, dist_der

    else:
        return p1_cruce, p2_cruce, dist_cruce



p1, p2, distancia = closest_pair_1d(LISTA_1D)

print("Divide y vencerás")
print("Puntos más cercanos:", p1, p2)
```

En el caso del algoritmo anterior, la complejidad sera de O(n log n). Esto es porque ordenar cuesta O(n log n) mientras que en la recursión copias niveles entonces en cada nivel copias n elementos y hay log n niveles por lo que n log n + n log n = O(n log n)
Este no se podría mejorar.

```
#AHORA DIVIDE Y VENCERÁS CON 2D
#COMPEJIDAD DE O(n log^2 n)
#EN ESTE CASO CUANDO NOS QUEDAMOS CON 3 O MENOS NÚMEROS UTILIZAMOS FUERZA BRUTA
def distancia2(p1, p2):
    return (p1[0] - p2[0])**2 + (p1[1] - p2[1])**2


def fuerza_bruta(lista):
    min_dist = float("inf")
    p1 = None
    p2 = None

    for i in range(len(lista)):
        for j in range(i + 1, len(lista)):
            d = distancia2(lista[i], lista[j])

            if d == 0:
                continue

            if d < min_dist:
                min_dist = d
                p1 = lista[i]
                p2 = lista[j]

    return p1, p2, min_dist


def closest_pair_2d(lista):
    n = len(lista)

    if n <= 3:
        return fuerza_bruta(lista)

    mitad = n // 2
    x_mitad = lista[mitad][0]

    izquierda = lista[:mitad]
    derecha = lista[mitad:]

    p1_izq, p2_izq, dist_izq = closest_pair_2d(izquierda)
    p1_der, p2_der, dist_der = closest_pair_2d(derecha)

    if dist_izq <= dist_der:
        mejor_p1, mejor_p2, mejor_dist = p1_izq, p2_izq, dist_izq
    else:
        mejor_p1, mejor_p2, mejor_dist = p1_der, p2_der, dist_der

    # Zona central: puntos cerca de la línea de división
    banda = []

    for punto in lista:
        if (punto[0] - x_mitad) ** 2 < mejor_dist:
            banda.append(punto)

    # Ordenar la banda por coordenada y
    banda.sort(key=lambda p: p[1])

    # Comparar puntos cercanos dentro de la banda
    for i in range(len(banda)):
        for j in range(i + 1, len(banda)):
            if (banda[j][1] - banda[i][1]) ** 2 >= mejor_dist:
                break

            d = distancia2(banda[i], banda[j])

            if d == 0:
                continue

            if d < mejor_dist:
                mejor_dist = d
                mejor_p1 = banda[i]
                mejor_p2 = banda[j]

    return mejor_p1, mejor_p2, mejor_dist


# Ordenar por x antes de empezar
LISTA_2D.sort(key=lambda p: p[0])

p1, p2, distancia = closest_pair_2d(LISTA_2D)

print("Divide y vencerás")
print("Puntos más cercanos:", p1, p2)
```

```
#FUERZA BRUTA CON 3D
#COMPLEJIDAD DE O(n^2)
import random

LISTA_3D = [(random.randrange(1,10000),random.randrange(1,10000),random.randrange(1,10000)) for x in range(1000)]

def distancia2(p1, p2):
    return (
        (p1[0] - p2[0])**2 +
        (p1[1] - p2[1])**2 +
        (p1[2] - p2[2])**2
    )

min_dist = float("inf")
punto1 = None
punto2 = None

for i in range(len(LISTA_3D)):
    for j in range(i + 1, len(LISTA_3D)):
        d = distancia2(LISTA_3D[i], LISTA_3D[j])

        if d == 0:
            continue

        if d < min_dist:
            min_dist = d
            punto1 = LISTA_3D[i]
            punto2 = LISTA_3D[j]

print("Fuerza bruta")
print("Puntos más cercanos:", punto1, punto2)
```

```
#DIVIDE Y VENCERÁS CON 3D
#COMPLEJIDAD O(n log^2 n)

def distancia2(p1, p2):
    return (
        (p1[0] - p2[0])**2 +
        (p1[1] - p2[1])**2 +
        (p1[2] - p2[2])**2
    )

def closest_pair_3d(lista):
    n = len(lista)

    if n < 2:
        return None, None, float("inf")

    if n == 2:
        d = distancia2(lista[0], lista[1])

        if d == 0:
            return None, None, float("inf")

        return lista[0], lista[1], d

    mitad = n // 2
    x_mitad = lista[mitad][0]

    izquierda = lista[:mitad]
    derecha = lista[mitad:]

    p1_izq, p2_izq, dist_izq = closest_pair_3d(izquierda)
    p1_der, p2_der, dist_der = closest_pair_3d(derecha)

    if dist_izq <= dist_der:
        mejor_p1, mejor_p2, mejor_dist = p1_izq, p2_izq, dist_izq
    else:
        mejor_p1, mejor_p2, mejor_dist = p1_der, p2_der, dist_der

    banda = []

    for punto in lista:
        if (punto[0] - x_mitad) ** 2 < mejor_dist:
            banda.append(punto)

    banda.sort(key=lambda p: p[1])

    for i in range(len(banda)):
        j = i + 1

        while j < len(banda) and (banda[j][1] - banda[i][1]) ** 2 < mejor_dist:
            if (banda[j][2] - banda[i][2]) ** 2 < mejor_dist:
                d = distancia2(banda[i], banda[j])

                if d != 0 and d < mejor_dist:
                    mejor_dist = d
                    mejor_p1 = banda[i]
                    mejor_p2 = banda[j]

            j += 1

    return mejor_p1, mejor_p2, mejor_dist


LISTA_3D.sort(key=lambda p: p[0])

p1, p2, distancia = closest_pair_3d(LISTA_3D)

print("Divide y vencerás")
print("Puntos más cercanos:", p1, p2)
```
